```python
from litellm import completion
from typing import List, Dict


def generate_response(messages: List[Dict]) -> str:
    """Call LLM to get response"""
    response = completion(
        model="gemini/gemini-2.5-flash",
        messages=messages,
        max_tokens=1024
    )
    return response.choices[0].message.content


messages = [
    {"role": "system", "content": "You are an expert software engineer that prefers functional programming."},
    {"role": "user", "content": "Write a function to swap the keys and values in a dictionary."}
]

response = generate_response(messages)
print(response)
```

```python
import litellm

# The model name must be prefixed with 'gemini/' for LiteLLM to route correctly.
# Common models are: 'gemini/gemini-2.5-flash', 'gemini/gemini-2.5-pro', etc.

try:
    response = litellm.completion(
        # --- BARE MINIMUM ADAPTATION START ---
        model="gemini/gemini-2.5-flash", # Use a Gemini model name
        # --- BARE MINIMUM ADAPTATION END ---
        messages=[
            {"role": "user", "content": "Explain the importance of the environment variable GEMINI_API_KEY."}
        ],
        temperature=0.7
    )

    # Accessing the content is the same as the OpenAI example
    print(response.choices[0].message.content)

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
#
# Set your API key as shown in the Google Colab notebook
#

from litellm import completion
from typing import List, Dict

def generate_response(messages: List[Dict]) -> str:
    """Call LLM to get response"""
    response = completion(
        model="gemini/gemini-2.5-flash", # Use a Gemini model name
        messages=messages,
        temperature=0.3,
        max_tokens=4096 # 8192 is the max
    )
    return response.choices[0].message.content


messages = [
    {"role": "system", "content": "You are an expert software engineer that prefers functional programming."},
    {"role": "user", "content": "Write a function to swap the keys and values in a dictionary."}
]

response = generate_response(messages)
print(response)
```

```python
#
# Set your API key as shown in the Google Colab notebook
#

import base64
from litellm import completion
from typing import List, Dict

def generate_response_base64(messages: List[Dict]) -> str:
    """Calls LLM and returns the response content encoded in Base64."""

    # 1. Call LiteLLM to get the normal text response
    response = completion(
        model="gemini/gemini-2.5-flash", # Use a Gemini model name
        messages=messages,
        temperature=0.3,
        max_tokens=4096 # 8192 is the max
    )

    # 2. Extract the plain text content
    plain_text = response.choices[0].message.content

    # 3. Encode the plain text to Base64
    # The string must first be converted to bytes using .encode('utf-8')
    base64_bytes = base64.b64encode(plain_text.encode('utf-8'))

    # 4. Decode the bytes to a final Base64 string for printing/transmission
    base64_string = base64_bytes.decode('utf-8')

    return base64_string

messages = [
    {"role": "system", "content": "You are an expert software engineer that prefers functional programming."},
    {"role": "user", "content": "Write a function to swap the keys and values in a dictionary."}
]
response_base64 = generate_response_base64(messages)
print(response_base64)
```
